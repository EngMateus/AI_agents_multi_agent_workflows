```python
from langchain_text_splitters import (
    HTMLHeaderTextSplitter
)
```

```python
### Essa classe chunkeriza paginas web a nivel das tags dela e adiciona metadados para cada cabeçalho


html_string = """
<!DOCTYPE html>
<html>
<body>
    <h1>Visão Geral do Projeto</h1>
    <p>Este é o parágrafo de introdução sobre o projeto de IA.</p>
    <h2>Instalação</h2>
    <p>Use o pip para instalar as dependências necessárias.</p>
    <h3>Dependências</h3>
    <p>LangChain, OpenAI e FAISS.</p>
</body>
</html>
"""

# defino aonde o texto deve ser cortado e os metadados a ser incluido
headers_to_slipt = [
    ("h1","Header 1"),
    ("h2","Header 2"),
    ("h3","Header 3")
] 

html_splitter = HTMLHeaderTextSplitter(
    headers_to_split_on=headers_to_slipt
)

html_header_splits = html_splitter.split_text(
    html_string
)


for idx,split in enumerate(html_header_splits):
    print(f"Índice [{idx}]:")
    print(f"  Conteúdo: {split.page_content}")
    print(f"  Metadados: {split.metadata}\n")
```
