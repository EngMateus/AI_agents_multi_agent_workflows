### TEXT SPLITTING FROM DOCUMENTS

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_text_splitters import CharacterTextSplitter
```

```python
document = PyPDFLoader(
    file_path=r"C:\Users\Mateus\Desktop\Udemy\Complete Agentic AI Bootcamp\Secao5 - IngestãoDados\Documentos\Computer-Vision-Resources.pdf",
)

document_loader = document.load()

print(f'Numero de paginas -> {len(document_loader)}',end='\n\n')

for page in range(0,15):
    print(document_loader[page].page_content)
```

```python
Splitter = RecursiveCharacterTextSplitter(
    chunk_size = 500, # cada bloco terá até 500 caracteres
    chunk_overlap = 100, # chunk overlap é a sobreposição de uma parte do bloco anterior no atual
    separators=["\n\n", "\n", " ", ""]
)


chunks = Splitter.split_documents(
    document_loader
)

print(f'Quantidade de chunks gerados -> {len(chunks)}',end='\n\n')

for i,chunk in enumerate(chunks):
    print(f'Chunk numero -> {i}')
    print(chunk.page_content)
    print("********************************************")
```

```python
### PARA ARQUIVOS OD TIPO TXT, É IMPORTANTE CONVERTER-LOS EM DOCUMENT ANTES DE ENVIAR PARA UM BANCO VETORIAL

txt_document = TextLoader(
    file_path=r'C:\Users\Mateus\Desktop\Udemy\Complete Agentic AI Bootcamp\Secao5 - IngestãoDados\Documentos\Ai_Agents.txt.txt',
    encoding="utf-8"
)

docs = txt_document.load()

split = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

txt_splitado = split.split_documents(docs)

for i,chun in enumerate(txt_splitado):
    print(f'Chunk -> {i}')
    print(chun.page_content)
```

###  Funcionamento do Algoritmo `RecursiveCharacterTextSplitter`

1. **Divisão Inicial (`\n\n`):** O algoritmo tenta primeiro dividir o texto em blocos utilizando parágrafos ou quebras de linha duplas.
2. **Verificação de Tamanho:** Ele analisa se algum desses blocos isolados excede o limite definido em `chunk_size` (ex: 500 caracteres).

---

###  Comportamento de Divisão e Overlap

* **Quando o bloco é MENOR que o `chunk_size`:**
  O algoritmo **não divide** o bloco e mantém o parágrafo inteiro. Como o texto não foi cortado, **não há necessidade nem aplicação de `chunk_overlap`**.

* **Quando o bloco é MAIOR que o `chunk_size`:**
  O algoritmo é forçado a fatiar o texto utilizando os próximos delimitadores (ex: `\n`, `" "`). É **apenas nesse momento** que o `chunk_overlap` é aplicado, repetindo o trecho final do bloco anterior no início do próximo para manter a continuidade do contexto.

### ⚠️ Observação sobre o `CharacterTextSplitter`

Embora seja abordado em muitos cursos e tutoriais, o **`CharacterTextSplitter`** é considerado uma abordagem engessada e praticamente obsoleta para pipelines de RAG modernos.

* **Divisão Rígida:** Ele tenta dividir o texto utilizando apenas **um único separador estático** (por padrão `\n\n`). 
* **Perda de Contexto Semântico:** Se um bloco de texto ultrapassar o `chunk_size` e não encontrar o separador configurado, ele é forçado a cortar o texto de forma arbitrária, correndo o risco de quebrar frases e parágrafos ao meio.
* **Boa Prática:** Em projetos reais e na linguagem natural, prefere-se o uso do **`RecursiveCharacterTextSplitter`**, que opera com múltiplos separadores hierárquicos e preserva a estrutura e o sentido do documento.
