## Pydantic Basics: Creating and Using Models

Pydantic models are the foundation of data validation in Python. They use Python type annotations to define the structure of data and validate it automatically at runtime.

By defining fields with explicit types, Pydantic ensures that incoming data conforms to the expected schema, helping prevent invalid values and reducing errors in application code.

```python
from pydantic import BaseModel
from dataclasses import dataclass


## Classe normal do python, sem validação de dados
## nessa simples classe, definimos os atributos dela e seus annotations
@dataclass
class Person():
    name:str
    age:int
    city:str
    

## aqui eu segui as regras dos tipos de dados que a minha classe foi construida
## se o nome foi definido como str, eu passei uma string, se a idade foi definida com int, passei numero...
## aqui, posso consider um mundo ideal de passagem de dados
person1 = Person(
    name='Mateus',
    age=28,
    city='Toledo'
)

## nessa instancia, ao invez de seguir o type annotation conforme a classe foi construida
## eu no lugar de passar uma string na var 'city' eu passei um numero
## o objetivo aqui é verificar se a classe aceita tipo diferente de dados 
person2 = Person(
    name='Mateus',
    age=28,
    city=2026
)



print(f"""
      Atributos da Pessoa1
      Tipo instancia -> {type(person1)}
      Tipo nome -> {type(person1.name)}
      Tipo idade -> {type(person1.age)}
      Tipo cidade -> {type(person1.city)}
      """
)


print(f"""
      Atributos da Pessoa2
      Tipo instancia -> {type(person2)}
      Tipo nome -> {type(person2.name)}
      Tipo idade -> {type(person2.age)}
      Tipo cidade -> {type(person2.city)}
      """
)


print("""
      Como visto na instancia person2, o atributo cidade aceitou eu passar int, sendo que ela foi definida com annotation do tipo str
      Ou seja, não houve uma real validação nos dados 
      É aqui que entra o pydatinc
      """)
```

## Pydantic Validation with `BaseModel`

Now we will test the same model using `BaseModel` to verify how Pydantic validates data at runtime.

With `BaseModel`, the declared field types are enforced automatically, so invalid values raise validation errors instead of being accepted silently.


```python
### detalhe que, se no atriburo year que esta definido como int, e eu colocar '2020' que é string
### o pydantic automaticamente vai converter de string para int, e não vai lançar erro de validação
## caso eu colocar no lugar qualquer string que não tenha somente numeros inteiros, dai ele lança erro

from pydantic import BaseModel

class Car(BaseModel):
    model:str
    year:int
    price:float
    
    
# seguindo os annotations conforme a classe pede
car1 = Car(
    model='Del Rey',
    year=1985,
    price=12.589
)

car2 = Car(
    model='Fiat Uno',
    year='2025a',
    price='10.000' 
)

print(f"""
      Atributos Car1
      Tipo modelo -> {type(car1.model)}
      Tipo ano -> {type(car1.year)}
      Tipo preço -> {type(car1.price)}
""")



print(f"""
      Atributos car2
      Tipo modelo -> {type(car2.model)}
      Tipo ano -> {type(car2.year)}
      Tipo preço -> {type(car2.price)}
""")
```

### Classes with Optional Field

```python

## ao usarmos o Optional -> indicamos ao python e ao pydantic que esse campo pode ser nulo ou do tipo especificado (float,bool) nos casos definidos abaixo
## quando None, não é obrigatório ser preenchido
## quando instanciar um objeto e não definir o salary por exemplo, o valor padrão dele vai ser None
## quando instanciar um objeto e não definir o is_active, o valor padrão vai ser True
## usado BaseModel para validação dos dados

from pydantic import BaseModel
from typing import Optional

class Employee(BaseModel):
    id:int
    name:str
    departament:str
    salary:Optional[float] = None # Optional with default Value
    is_active:Optional[bool] = True # Optional with default True
    
    
    
    
employee1 = Employee(
    id=10,
    name='Mateus Silva',
    departament='Ai Agents engineer'
)

print(f"""
      ID -> {employee1.id}
      Name -> {employee1.name}
      Departament -> {employee1.departament}
      Salary -> {employee1.salary}
      Active? -> {employee1.is_active}
      """)
```

```python
from pydantic import BaseModel
from typing import List


class Classroom(BaseModel):
    room_number:str
    students:List[str] # list of strings
    capacity:int
    
    
class_room1 = Classroom(
    room_number = '10a',
    students = ['João','Maria','José'],
    capacity = 3
)

print(class_room1)
```

```python
from pydantic import(
    BaseModel,
    Field
)

### alem da validação dos tipos de dados que o pydantic faz
### Ao usar o Field, dou mais um nivel de validação de dados
### min_lenght -> numero minimo de caracteres no atributo name por exemplo
### max_lenght -> numero maximo de caracteres no atributo name por exemplo
### gt -> greater than ;;;;;; le -> less than == greater than 0 and less than 100000

class item(BaseModel):
    name:str = Field(min_length=5,max_length=15)
    price:float = Field(gt=0,le=1000)
    quantity:int = Field(gt=0)
    
    
    
item1 = item(
    name = 'Nvidea RTX 5090',
    price = 500,
    quantity=1
)

print(item1,end='\n\n\n')


try:
    item2 = item(
    name = 'Corsair Mechanic KeyBoard',
    price = 5000,
    quantity=0
    )
    
except ValueError as e:
    print('*****Erro lançado*****',end='\n\n')
    print(e)


```

```python
from pydantic import BaseModel, Field

### ... indica que esse atributo é obrigatorio na hora de instanciar
### os outros atributos possuem valores default, então, se eu não passar, o default vai assumir 
### no campo email default_factory=lambda:user@example.com -> fica dinamico a nivel de instancia
class User(BaseModel):
    username:str = Field(...,description="unique username for the user")
    age:int = Field(default=18,description="User age, defaults to 18")
    email:str = Field(default_factory=lambda: "user@example.com",description="Default email address")
    
    
user1 = User(username="Alice")
print(user1)
```

```python
print(User.model_json_schema())
```

```python
import time
from datetime import datetime
from pydantic import BaseModel, Field

class LogEntry(BaseModel):
    message: str
    # Executa a função no momento exato em que o log é instanciado
    created_at: datetime = Field(default_factory=lambda: datetime.now())
    
    
debug1 = LogEntry(
    message="Estanciado!"
)

time.sleep(3)

debug2 = LogEntry(
    message="Estanciado!"
)

print(debug1)

print(debug2)
```
