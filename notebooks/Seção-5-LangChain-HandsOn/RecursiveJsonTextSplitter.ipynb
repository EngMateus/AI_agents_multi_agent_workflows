```python
from langchain_text_splitters.json import RecursiveJsonSplitter
import requests
import json
```

```python
json_data = requests.get("https://pokeapi.co/api/v2/pokemon/ditto").json()

# json_data

spliter_json = RecursiveJsonSplitter(
  max_chunk_size = 300,
  min_chunk_size = None
)


json_chunks = spliter_json.split_json(json_data)

```

```python
json_chunks
```

```python
for chunk in json_chunks[:1]:
    print(chunk)
```

```python
### convertendo em documento 
json_to_docs = spliter_json.create_documents(texts=[json_data])
for chunk in json_to_docs[:2]:
    print(chunk)
    print('**********')
```
